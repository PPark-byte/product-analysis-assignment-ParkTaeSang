### 분석 문제와 공통 기준

이번 실습은 **2024년 7~9월 가입한 전체 산업의 신규 계정**을 대상으로 합니다. 기존 파일의 DevTools 구독 상태 비교를, 교안의 가입 시점·시간순 행동 분석에 맞게 변경했습니다. 별도의 1강 답안 없이도 풀 수 있도록 문제와 가설을 아래에 제공합니다.

- 분석 문제: 신규 계정은 가입한 달에 기능 사용까지 얼마나 도달하며, 이후에도 사용하는가?
- 목표 지표: 가입 월 기능 사용 도달률 = 가입 월에 실제 사용한 계정 수 / 대상 가입 계정 수.
- 초기 가설: 가입→구독 시작보다 구독 시작→기능 사용 구간의 전환율이 낮을 것이다. 이후 기능 사용률은 매월 감소할 것이다. **분석 전 가설이며 실제 결과와 다를 수 있습니다.**
- 분석 단위는 `account_id`입니다. 한 계정의 여러 구독·사용 기록은 중복 계정으로 세지 않습니다.
- 실제 사용은 `usage_count > 0`, 사용일이 해당 구독 시작일 이상이며 종료일이 있으면 종료일 이하인 기록입니다. 가입 전 시작된 구독은 이 신규 가입 경로에서 제외합니다.
- 달은 달력 월입니다. 가입 월 말까지를 퍼널 관찰 기간으로 사용하므로 월말 가입자는 관찰 일수가 짧습니다. 동일한 가입 후 30일 지표와는 다릅니다.
- 파일의 최종 사용일을 분석 종료일로 사용하며, 그때까지의 로그가 제공되었다고 가정합니다. 로그가 없다는 것은 기록상 미사용이며 해지·영구 이탈을 뜻하지 않습니다.

| feature_usage: subscription_id, usage_date, usage_count | 실제 기능 사용과 사용 월 확인 |

## 실습 준비

아래 코드는 공통으로 제공합니다. 데이터 불러오기, 분석 컬럼·결측치·자료형, 기초 통계량·분포를 확인하고 분석용 표를 준비하세요. 테이블 연결은 Ravenstack 구조에 맞춘 준비 과정이며 새로운 통계 기법을 요구하지 않습니다.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = Path(".")
folders = [DATA_DIR, Path("upload"), Path("data"), Path("/content")]
data = {}
for name in ["accounts", "subscriptions", "feature_usage"]:
    names = [f"ravenstack_{name}(2).csv", f"ravenstack_{name}.csv"]
    path = next((d/f for f in names for d in folders if (d/f).exists()), None)
    if path is None:
        raise FileNotFoundError(f"{name} CSV를 올리거나 DATA_DIR을 수정하세요.")
    data[name] = pd.read_csv(path)
    print(path.name, data[name].shape)

accounts = data["accounts"][["account_id", "signup_date"]].copy()
subscriptions = data["subscriptions"][["subscription_id", "account_id", "start_date", "end_date"]].copy()
usage = data["feature_usage"][["subscription_id", "usage_date", "usage_count"]].copy()
for frame, cols in [(accounts, ["signup_date"]), (subscriptions, ["start_date", "end_date"]), (usage, ["usage_date"])]:
    for col in cols:
        frame[col] = pd.to_datetime(frame[col], errors="raise")
    print("\n분석 컬럼 예시:\n", frame.head(3).to_string(index=False))
    print("자료형:\n", frame.dtypes.to_string())
    print("결측치:\n", frame.isna().sum().to_string())
# end_date 결측은 종료일이 기록되지 않은 구독이므로 삭제하지 않습니다.
assert accounts.account_id.is_unique
assert subscriptions.subscription_id.is_unique
assert not accounts.isna().any().any()
assert not subscriptions.drop(columns="end_date").isna().any().any()
assert not usage.isna().any().any()
print("\n사용 횟수 기초 통계량:\n", usage.usage_count.describe().to_string())
print("가입 월별 계정 수:\n", accounts.groupby(accounts.signup_date.dt.to_period("M")).size().to_string())
analysis_end = usage.usage_date.max()
print("사용 기록 범위:", usage.usage_date.min().date(), "~", analysis_end.date())

target = accounts.loc[accounts.signup_date.between("2024-07-01", "2024-09-30")].copy()
target["cohort"] = target.signup_date.dt.to_period("M")
subs = subscriptions.merge(target, on="account_id", how="inner", validate="many_to_one")
excluded_subs = (subs.start_date < subs.signup_date).sum()
subs = subs.loc[subs.start_date >= subs.signup_date].copy()
events = usage.merge(subs, on="subscription_id", how="inner", validate="many_to_one")
valid = ((events.usage_count > 0) & (events.usage_date >= events.start_date)
         & (events.end_date.isna() | (events.usage_date <= events.end_date)))
excluded_events = (~valid).sum()
events = events.loc[valid].copy()
events["month_offset"] = ((events.usage_date.dt.year - events.signup_date.dt.year) * 12
                          + events.usage_date.dt.month - events.signup_date.dt.month)
print("\n대상 신규 계정 수:", len(target))
print("가입 전 시작으로 제외한 구독 수:", excluded_subs)
print("사용 조건을 충족하지 않아 제외한 연결 기록 수:", excluded_events)
print("코호트별 가입 계정 수:\n", target.groupby("cohort").size().to_string())
```

## 필수 1 — 신규 계정의 가입 월 퍼널 분석

### 상황

가입한 달에 기능 사용까지 도달하지 못하는 구간을 찾으려고 합니다. 실제 데이터에 없는 화면 조회·온보딩 이벤트는 만들지 않습니다.

### 수행 사항

1. `target`, `subs`, `events`로 아래 세 단계의 고유 계정 수를 계산하세요.
   - 가입: 대상 계정 전체
   - 구독 시작: 가입일 이후이면서 가입 월 안에 구독이 시작된 계정
   - 기능 사용: 해당 구독에서 가입 월 안에 실제 사용한 계정
2. 이전 단계 대비 전환율·이탈률과 미도달 계정 수를 표로 만드세요. 첫 단계의 전환율·이탈률은 비교 대상이 없어 빈 값으로 둡니다.
3. 막대그래프로 단계별 계정 수를 표시하고 가장 낮은 전환율 구간을 찾으세요.
4. 가입 월 기능 사용 도달률도 계산하여 초기 가설과 대조하세요.

### 질문

Q1. 세 단계의 계정 수와 구간별 전환율·이탈률·미도달 수는 얼마인가요?

Q2. 가장 낮은 전환율 구간은 어디이며, 퍼널 초기 가설과 일치하나요?

Q3. 가입 월 기능 사용 도달률은 얼마이며, 마지막 구간 전환율과 분모가 어떻게 다른가요?

Q4. 왜 계정 수를 중복 제거해야 하며, 이 결과만으로 미사용 원인을 확정할 수 있나요?

### 체크 포인트

같은 분석 단위, 시간 순서, 이전 단계 분모, 데이터로 확인한 사실과 원인 가설의 구분.


```python
# 필수 1 풀이 코드를 작성하세요.
# 준비 코드의 target, subs, events를 활용하세요.

# 1) 가입 월 안에 시작된 구독만 고르기
# 구독 시작 월이 가입 월과 같으면 가입 월 안에 시작한 것
subs["start_month"] = subs["start_date"].dt.to_period("M")
subs_m0 = subs[subs["start_month"] == subs["cohort"]]
print("가입 월 안에 시작된 구독 수:", len(subs_m0))

# 2) 가입 월 안에 사용한 기록만 고르기 (가입 월 안에 시작된 구독에서)
events["usage_month"] = events["usage_date"].dt.to_period("M")
events_m0 = events[(events["month_offset"] == 0) & (events["subscription_id"].isin(subs_m0["subscription_id"]))]
print("가입 월 안의 실제 사용 기록 수:", len(events_m0))

# 3) 단계별 고유 계정 수 (nunique로 중복 제거)
step1 = target["account_id"].nunique()
step2 = subs_m0["account_id"].nunique()
step3 = events_m0["account_id"].nunique()
print("가입 계정 수:", step1)
print("구독 시작 계정 수:", step2)
print("기능 사용 계정 수:", step3)

# 4) 퍼널 표 만들기
funnel = pd.DataFrame({
    "단계": ["가입", "구독 시작", "기능 사용"],
    "계정 수": [step1, step2, step3]
})
funnel["이전 단계 계정 수"] = funnel["계정 수"].shift(1)
funnel["전환율(%)"] = (funnel["계정 수"] / funnel["이전 단계 계정 수"] * 100).round(1)
funnel["이탈률(%)"] = (100 - funnel["전환율(%)"]).round(1)
funnel["미도달 계정 수"] = funnel["이전 단계 계정 수"] - funnel["계정 수"]
print("\n[퍼널 표]")
print(funnel)

# 5) 막대그래프
plt.rcParams["font.family"] = "Malgun Gothic"   # 한글 깨짐 방지 (맥이면 "AppleGothic"으로 바꾸기)
plt.rcParams["axes.unicode_minus"] = False
plt.figure(figsize=(6, 4))
plt.bar(funnel["단계"], funnel["계정 수"])
for i, v in enumerate(funnel["계정 수"]):
    plt.text(i, v, str(v), ha="center", va="bottom")
plt.title("2024년 7~9월 가입 계정 가입 월 퍼널")
plt.ylabel("계정 수")
plt.show()

# 6) 가장 낮은 전환율 구간 찾기
low_idx = funnel["전환율(%)"].idxmin()
print("\n가장 낮은 전환율 구간:", funnel.loc[low_idx - 1, "단계"], "→", funnel.loc[low_idx, "단계"],
      "/", funnel.loc[low_idx, "전환율(%)"], "%")

# 7) 가입 월 기능 사용 도달률 (분모 = 가입 계정 전체)
reach_rate = step3 / step1 * 100
print("가입 월 기능 사용 도달률:", round(reach_rate, 1), "%  (", step3, "/", step1, ")")
print("마지막 구간 전환율:", round(step3 / step2 * 100, 1), "%  (", step3, "/", step2, ")")
```

### 체크 포인트

가입 월 기준 코호트, 고정 분모와 중복 제거, 관찰 기간 확인, 그래프와 일치하는 해석.


```python
# 필수 2 풀이 코드를 작성하세요.
# 준비 코드의 target, subs, events를 활용하세요.

# 1) 관찰 기간 확인: 코호트별 M3 월말이 분석 종료일(analysis_end) 이전인지
print("분석 종료일:", analysis_end.date())
for c in sorted(target["cohort"].unique()):
    m3_month = c + 3   # 가입 월 + 3개월
    ok = m3_month <= analysis_end.to_period("M")   # 월 단위로 비교
    print(c, "코호트 M3 월:", m3_month, "/ 관찰 가능:", ok)

# 2) 코호트별 가입 계정 수 (분모 고정)
cohort_size = target.groupby("cohort")["account_id"].nunique()
print("\n코호트별 가입 계정 수:")
print(cohort_size)

# 3) M1~M3 월별 실제 사용 계정 수 (중복 제거)
events_m13 = events[events["month_offset"].between(1, 3)]
use_cnt = events_m13.groupby(["cohort", "month_offset"])["account_id"].nunique().unstack()
use_cnt = use_cnt.fillna(0).astype(int)   # 모두 관찰 가능한 달이라 0으로 채워도 됨
use_cnt.columns = ["M1", "M2", "M3"]
print("\n[월별 사용 계정 수]")
print(use_cnt)

# 4) 리텐션 표 (M0 = 100%)
retention = pd.DataFrame(index=cohort_size.index)
retention["M0"] = 100.0
for col in ["M1", "M2", "M3"]:
    retention[col] = (use_cnt[col] / cohort_size * 100).round(1)
print("\n[리텐션 표 (%)]")
print(retention)

# 5) 선그래프
plt.figure(figsize=(6, 4))
for c in retention.index:
    plt.plot(["M0", "M1", "M2", "M3"], retention.loc[c], marker="o", label=str(c))
plt.title("가입 월 코호트별 기능 사용 리텐션")
plt.ylabel("리텐션(%)")
plt.legend()
plt.show()

# 6) M3 최고 / 최저 코호트
best = retention["M3"].idxmax()
worst = retention["M3"].idxmin()
print("\nM3 최고 코호트:", best, retention.loc[best, "M3"], "%")
print("M3 최저 코호트:", worst, retention.loc[worst, "M3"], "%")
print("차이:", round(retention.loc[best, "M3"] - retention.loc[worst, "M3"], 1), "%p")

# 7) 변화폭 (%p)
change = pd.DataFrame(index=retention.index)
change["M1→M2(%p)"] = (retention["M2"] - retention["M1"]).round(1)
change["M2→M3(%p)"] = (retention["M3"] - retention["M2"]).round(1)
print("\n[변화폭]")
print(change)
```

### 제출물 / 체크 포인트

과제의 계산 코드·출력, 네 기준 비교표, 퍼널·리텐션 수치에 근거한 우선순위와 AARRR 단계, 원인 가설을 포함한 네 항목의 개선 시나리오, 가설이 지지되지 않을 때의 대안 또는 실행 제약 조건, Q1~Q5 답변을 제출합니다. 필수 1·2는 연습용이며, 과제 판단에 사용한 분석 수치는 과제 답안에도 명시하세요. 정답의 우선순위와 달라도 계산과 목표 지표 연결이 타당하고 검증 가능한 시나리오이면 인정합니다.

```python
# 과제 1 풀이 코드를 작성하세요.
# 준비 코드의 target, subs, events를 활용하세요.
# 필수 1의 subs_m0, events_m0 / 필수 2의 worst를 그대로 사용하니 위 셀을 먼저 실행하세요.

total = target["account_id"].nunique()
print("전체 대상 가입 계정 수:", total)

# 필수 1에서 만든 계정 목록 다시 준비
all_acc = set(target["account_id"])
sub_acc = set(subs_m0["account_id"])        # 가입 월 구독 시작 계정
use_acc = set(events_m0["account_id"])      # 가입 월 기능 사용 계정

# A: 가입 월에 구독을 시작하지 않은 계정
A = all_acc - sub_acc
# B: 가입 월에 구독은 시작했지만 가입 월 기능 사용까지 못 간 계정
B = sub_acc - use_acc
# C: M3 리텐션 최저 코호트 중 M3에 사용하지 않은 계정
worst_acc = set(target[target["cohort"] == worst]["account_id"])
m3_acc = set(events[(events["cohort"] == worst) & (events["month_offset"] == 3)]["account_id"])
C = worst_acc - m3_acc

cand = pd.DataFrame({
    "후보": ["A: 가입 월 구독 미시작", "B: 구독 시작 후 가입 월 미사용", f"C: {worst} 코호트 M3 미사용"],
    "영향 계정 수": [len(A), len(B), len(C)]
})
cand["전체 대비 비율(%)"] = (cand["영향 계정 수"] / total * 100).round(1)
print("\n[후보별 영향 범위]")
print(cand)

# 후보끼리 겹치는지 확인 (그래서 비율을 더하면 안 됨)
print("\nA와 C 겹치는 계정 수:", len(A & C))
print("B와 C 겹치는 계정 수:", len(B & C))

# 기대 효과 계산에 쓸 현재 값
print("\n현재 가입 월 기능 사용 도달률:", round(len(use_acc) / total * 100, 1), "%")
print("현재 구독 시작→기능 사용 전환율:", round(len(use_acc) / len(sub_acc) * 100, 1), "%")
# 가정: B 계정 중 10개가 추가로 가입 월 기능 사용에 도달한다면?
plus = 10
print("가정(B 중", plus, "개 추가 도달) 시 도달률:", round((len(use_acc) + plus) / total * 100, 1), "%")
print("가정 시 구독 시작→기능 사용 전환율:", round((len(use_acc) + plus) / len(sub_acc) * 100, 1), "%")

# 참고: 구독 시작일부터 가입 월 말까지 며칠이나 남았는지 (관찰 기간이 짧은지 확인)
first_start = subs_m0.groupby("account_id")["start_date"].min().reset_index()
first_start["월말"] = first_start["start_date"] + pd.offsets.MonthEnd(0)
first_start["남은 일수"] = (first_start["월말"] - first_start["start_date"]).dt.days
print("\n구독 시작일 ~ 가입 월 말까지 남은 일수 통계:")
print(first_start["남은 일수"].describe().round(1))
```

  - 비교 지표(양쪽 동일): 분자 = 가입 월 안에 실제 기능 사용(usage_count > 0)한 계정 수, 분모 = 그 그룹에 배정된 가입 계정 수, 관찰 기간 = 가입 월 말까지

Q4. 20.8%는 **관측 결과가 아니라 예상(가정)** 입니다. 실제로 관측한 값은 도달률 6.9%, 전환율 9.6%, 리텐션 수치들뿐입니다.
원인과 효과는 이렇게 구분했습니다. 데이터로 확인한 건 "구독 후 가입 월에 안 썼다"는 사실이고, "안내가 없어서"는 제 원인 가설입니다. 효과도 A/B 테스트에서 A와 B의 도달률 차이를 봐야 확인할 수 있는 거라, 지금은 원인도 효과도 확정이 아닙니다.

Q5. **대안**을 선택했습니다. B안이 A안보다 도달률이 안 오르면, 원인이 "안내 부족"이 아니라 "월말에 구독을 시작해서 관찰 기간이 짧은 것"일 수 있습니다(남은 일수 중앙값 8일). 그래서 다음에는 ① 가입 월 기준 말고 "구독 시작 후 14일 안에 사용한 비율"로 다시 계산해서 월말 가입 영향을 빼고 비교해 보고, ② feature_usage의 error_count를 확인해서 첫 사용 때 오류가 많은지 볼 것입니다. 그래도 원인이 안 보이면 우선순위 2번인 A(구독 시작 유도)로 넘어가겠습니다.
